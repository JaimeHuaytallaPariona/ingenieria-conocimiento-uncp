*REGRESIÓN SIMPLE*

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
x=np.array([1,1.5,2,2.5,3,3.5,4,4.5,5,5.5,6,6.5])
y=np.array([1.3,1.8,2.2,2.6,3.1,3.8,4.2,4.4,4.9,5.7,6.3,6.8])
```

```python
#Modelo de Regresión Lineal usando Mínimos Cuadrados
from sklearn import linear_model

n=x.shape[0]
xs=x.reshape((n,1))
ys=y.reshape((n,1))

modelo_rl=linear_model.LinearRegression()
modelo_rl.fit(xs,ys)
y_pred=modelo_rl.predict(xs)

plt.plot(x,y,"o",color="b")
plt.plot(x,y_pred,"-",color="r")
plt.grid()
plt.axis("equal")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Regresión Lineal")
plt.show()
```

```python
print(modelo_rl.coef_)
print(modelo_rl.intercept_)
```

```python
#Modelo usando una neurona artificial

modelo_nn=tf.keras.Sequential([tf.keras.layers.Dense(units=1,input_shape=[1])])
modelo_nn.compile(optimizer="sgd",loss="mean_squared_error")
modelo_nn.fit(x,y,epochs=8)

```

```python
y_pred_nn=modelo_nn.predict(x)

plt.plot(x,y,"o",color="b")
plt.plot(x,y_pred,"-",color="r",label="Minimos Cuadrados")
plt.plot(x,y_pred_nn,"-",color="g",label="Redes Neuronales")
plt.legend()
plt.grid()
plt.axis("equal")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Regresión Lineal")
plt.show()
```

```python
modelo_nn.get_weights()
```

```python
#Prediccion

x_new=np.array([[7.0]])

yp_rl=modelo_rl.predict(x_new)
yp_nn=modelo_nn.predict(x_new)

print("La prediccion para {} con mínimos cuadrados es: {}".format(x_new,yp_rl))
print("La prediccion para {} con una neurona artificial es: {}".format(x_new,yp_nn))
```
